```python
import re

def clean_latex(text):
    """
    Remove LaTeX syntax while preserving meaningful words.
    """

    # Remove $
    text = re.sub(r"\$", "", text)

    # Replace common Greek letters
    greek_letters = {
        r"\\alpha": " alpha ",
        r"\\beta": " beta ",
        r"\\gamma": " gamma ",
        r"\\delta": " delta ",
        r"\\lambda": " lambda ",
        r"\\mu": " mu ",
        r"\\sigma": " sigma ",
        r"\\phi": " phi ",
        r"\\psi": " psi ",
        r"\\omega": " omega ",
        r"\\theta": " theta ",
        r"\\pi": " pi "
    }

    for latex, word in greek_letters.items():
        text = re.sub(latex, word, text)

    # Remove remaining backslashes
    text = re.sub(r"\\", " ", text)

    # Remove braces
    text = re.sub(r"[{}]", " ", text)

    # Replace underscore with space
    text = re.sub(r"_", " ", text)

    # Remove ^
    text = re.sub(r"\^", " ", text)

    # Remove multiple spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
import re

custom_stopwords = {
    "paper",
    "study",
    "results",
    "result",
    "present",
    "presented",
    "based",
    "find",
    "found",
    "new",
    "show",
    "shown",
    "using",
    "approach",
    "proposed",
    "propose"
}

def preprocess_ml(text):

    doc = nlp(text)

    tokens = []

    for token in doc:

        word = token.lemma_.lower()

        # Remove stopwords
        if token.is_stop:
            continue

        # Remove punctuation
        if token.is_punct:
            continue

        # Remove numbers
        if token.like_num:
            continue

        # Remove one-letter words
        if len(word) == 1:
            continue

        # Remove custom stopwords
        if word in custom_stopwords:
            continue

        tokens.append(word)

    return " ".join(tokens)
```

```python
df["text_ml"] = df["text_clean"].apply(preprocess_ml)
```
